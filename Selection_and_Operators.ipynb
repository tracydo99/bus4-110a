```python
HoursWorked = 50
HourlyRate = 20
if HoursWorked <= 40:
  Regular_pay = HoursWorked * HourlyRate
  OverTime_pay = 0
else:
  Regular_pay = 40 * HourlyRate
  OverTime_Hours = HoursWorked - 40
  OverTime_pay = OverTime_Hours * (1.5 * HourlyRate)

Total_pay = Regular_pay + OverTime_pay
print(Total_pay)
```

```python
wage = 20
gross = 0.0
hours = 50
hour_limit = 40

if hours <= hour_limit:
  gross = wage * hours
else:
  gross = hour_limit * wage
  gross += (hours - hour_limit) * wage * 1.5
print(gross)
```

```python
income = 30000 # Example value
if income <= 10000:
    tax = .05
elif income <= 20000:
    tax = .10
elif income < 50000: # Corrected condition
    tax = .20 # Tax rate for income > 20000 and < 50000
else: # Income >= 50000
    tax = .15 # Tax rate for income >= 50000

print(tax)
```
