*keyword arguments (kwargs)*: each argument consists of a variable name and a value, and lists and dictionaries of values, can be in any order
*positional arguments (args)*: need to be in the same order the parameters were written

```python
def AddNumbers(x,y,z):
  print(x+y+z)
AddNumbers(2,4,5)
```

```python
def describe_pet(animal_type, pet_name):
  print("My "+animal_type+" is named "+ pet_name.title()+".")
describe_pet('hamster','harry') #positional arg
describe_pet(animal_type='hamster',pet_name='harry') #keyword arg
```

```python
def get_formatted_name(first_name, last_name):
  full_name = first_name + ' ' + last_name
  return full_name.title()
print(get_formatted_name('jimi','hendrix'))
```

```python
def get_formatted_name(first_name, last_name, middle_name = ""):
  full_name = first_name + ' ' + middle_name + ' ' + last_name
  return full_name.title()
musician = get_formatted_name('Hollis','Frazier-Herndon','Parker')
print(musician)
```

default values always have to be on the right, then adjust positional args accordingly

```python
def greet_users(names):
  for name in names:
    msg = "Hey, " + name.title() + "!"
    print(msg)
usernames = ['tracy','khang','margot']
greet_users(usernames)
```

```python
def make_shirt(size, text):
  print('Your size '+size+" t-shirt will say, \""+ text +".\"")
make_shirt('small','I <3 SJSU') #positional arg
make_shirt(size='small',text='I <3 SJSU') #keyword arg
```
