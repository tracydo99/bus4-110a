```python
prompt = "I will sum all the numbers starting from n down to 1.\nGive me a value for n: "
response = int(input(prompt))
n = response
sum = 0
while n >= 1:
  sum += n
  n -= 1
print(sum)
```

```python
prompt = "I will calculate n!\nGive me a value for n: "
response = int(input(prompt))
x = 1
factorial = 1
while x <= response:
  factorial *= x
  x += 1
print(factorial)
```
